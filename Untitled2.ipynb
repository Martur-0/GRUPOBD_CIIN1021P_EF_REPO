```python
from pyspark.sql import SparkSession
import time


spark = SparkSession.builder.appName("ProyectoMINSA").getOrCreate()
```

```python

df = spark.read.csv("dataset_limpio_etapa.csv", header=True, inferSchema=True)
```

```python

df = spark.read.csv("dataset_limpio_etapa.csv", header=True, inferSchema=True, sep=";")


df.show(5)
```

```python
inicio = time.time()

resultado_df = df.groupBy("provincia").count()
resultado_df.show()

fin = time.time()
print(f"Tiempo usando DataFrame: {fin - inicio} segundos")
```

```python

df.createOrReplaceTempView("tabla_minsa")


inicio_sql = time.time()


resultado_sql = spark.sql("""
    SELECT provincia, COUNT(*) as Total_Atenciones
    FROM tabla_minsa
    GROUP BY provincia
    ORDER BY Total_Atenciones DESC
""")


resultado_sql.show()


fin_sql = time.time()
print(f"Tiempo usando SparkSQL: {fin_sql - inicio_sql} segundos")
```
